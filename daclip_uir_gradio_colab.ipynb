```python
import os
import re

# Define the path to the app.py file
app_py_path = '/content/daclip-uir/universal-image-restoration/config/daclip-sde/app.py'

# Read the content of app.py
with open(app_py_path, 'r') as f:
    app_py_content = f.read()

# Modify the line that calls open_clip.create_model_from_pretrained
# to include weights_only=False in the load_checkpoint call within the function
# We need to find the call to load_checkpoint in the open_clip library
# Looking at the traceback, the call is in open_clip/factory.py
factory_py_path = '/content/daclip-uir/universal-image-restoration/open_clip/factory.py'

with open(factory_py_path, 'r') as f:
    factory_py_content = f.read()

# Add weights_only=False to the torch.load call within load_state_dict
modified_factory_py_content = re.sub(
    r'checkpoint = torch.load\((.*?)\)',
    r'checkpoint = torch.load(\1, weights_only=False)',
    factory_py_content
)

# Write the modified content back to factory.py
with open(factory_py_path, 'w') as f:
    f.write(modified_factory_py_content)

print(f"Modified {factory_py_path} to include weights_only=False in torch.load.")

# Now, re-run the original cell
```
